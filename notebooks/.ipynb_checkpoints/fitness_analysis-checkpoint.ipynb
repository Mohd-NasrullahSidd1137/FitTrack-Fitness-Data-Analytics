```python
# Step 7.11 - Compare Calories

calories_comparison = daily_activity[
    ["Id", "ActivityDate", "Calories"]
].merge(
    daily_calories[
        ["Id", "ActivityDay", "Calories"]
    ],
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    suffixes=("_activity", "_daily")
)

calories_comparison["Calories_Match"] = (
    calories_comparison["Calories_activity"]
    ==
    calories_comparison["Calories_daily"]
)

print("Calories mismatches:",
      (~calories_comparison["Calories_Match"]).sum())
```

```python
# Step 7.12 - Compare Steps

steps_comparison = daily_activity[
    ["Id", "ActivityDate", "TotalSteps"]
].merge(
    daily_steps[
        ["Id", "ActivityDay", "StepTotal"]
    ],
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    suffixes=("_activity", "_daily")
)

steps_comparison["Steps_Match"] = (
    steps_comparison["TotalSteps"]
    ==
    steps_comparison["StepTotal"]
)

print(
    "Steps mismatches:",
    (~steps_comparison["Steps_Match"]).sum()
)
```

```python
# Step 7.13 - Compare Activity Intensities

intensity_columns = [
    "SedentaryMinutes",
    "LightlyActiveMinutes",
    "FairlyActiveMinutes",
    "VeryActiveMinutes",
    "SedentaryActiveDistance",
    "LightActiveDistance",
    "ModeratelyActiveDistance",
    "VeryActiveDistance"
]

intensity_comparison = daily_activity[
    ["Id", "ActivityDate"] + intensity_columns
].merge(
    daily_intensities[
        ["Id", "ActivityDay"] + intensity_columns
    ],
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    suffixes=("_activity", "_daily")
)

for column in intensity_columns:
    mismatch_count = (
        intensity_comparison[f"{column}_activity"]
        != intensity_comparison[f"{column}_daily"]
    ).sum()

    print(f"{column} mismatches: {mismatch_count}")
```

```python
# Step 7.65 - Daily Calorie Summary

minute_calories_narrow["ActivityDate"] = (
    minute_calories_narrow["ActivityMinute"].dt.date
)

daily_calorie_summary = (
    minute_calories_narrow
    .groupby(["Id", "ActivityDate"], as_index=False)["Calories"]
    .sum()
    .rename(columns={
        "Calories": "TotalDailyCalories"
    })
)

print("Daily Calorie Summary Shape:",
      daily_calorie_summary.shape)

print("\nFirst 10 Records:")
print(daily_calorie_summary.head(10))
```

```python
# Step 7.66 - Daily Calorie Statistics

print("Daily Calorie Statistics:\n")

print(
    daily_calorie_summary["TotalDailyCalories"].describe()
)
```

```python
# Overall Average Daily Calories

average_daily_calories = (
    daily_calorie_summary["TotalDailyCalories"].mean()
)

print(
    "\nAverage Daily Calories:",
    round(average_daily_calories, 2)
)
```

```python
# Step 7.67 - Daily Intensity + Steps Analysis

# Convert ActivityHour to datetime
hourly_intensities["ActivityHour"] = pd.to_datetime(
    hourly_intensities["ActivityHour"]
)

hourly_steps["ActivityHour"] = pd.to_datetime(
    hourly_steps["ActivityHour"]
)

# Create ActivityDate
hourly_intensities["ActivityDate"] = (
    hourly_intensities["ActivityHour"].dt.date
)

hourly_steps["ActivityDate"] = (
    hourly_steps["ActivityHour"].dt.date
)

# Daily Intensity Summary
daily_intensity = (
    hourly_intensities
    .groupby(["Id", "ActivityDate"], as_index=False)
    .agg(
        TotalIntensity=("TotalIntensity", "sum"),
        AverageIntensity=("AverageIntensity", "mean"),
        AverageCalculatedIntensity=(
            "CalculatedAverageIntensity", "mean"
        ),
        AverageIntensityDifference=(
            "IntensityDifference", "mean"
        )
    )
)

# Daily Steps Summary
daily_steps_summary = (
    hourly_steps
    .groupby(["Id", "ActivityDate"], as_index=False)
    .agg(
        TotalSteps=("StepTotal", "sum")
    )
)

print("Daily Intensity Shape:", daily_intensity.shape)
print("Daily Steps Shape:", daily_steps_summary.shape)

print("\nDaily Intensity:")
print(daily_intensity.head())

print("\nDaily Steps:")
print(daily_steps_summary.head())
```

```python
# Step 7.68 - Sleep Analysis

# Convert SleepDay to datetime
sleep_day["SleepDay"] = pd.to_datetime(
    sleep_day["SleepDay"]
)

# Check columns
print("Sleep Dataset Columns:")
print(sleep_day.columns.tolist())

# Daily sleep summary
daily_sleep_summary = (
    sleep_day
    .groupby(["Id", "SleepDay"], as_index=False)
    .agg(
        TotalSleepMinutes=("TotalMinutesAsleep", "sum"),
        TotalTimeInBed=("TotalTimeInBed", "sum")
    )
)

# Convert minutes to hours
daily_sleep_summary["SleepHours"] = (
    daily_sleep_summary["TotalSleepMinutes"] / 60
)

daily_sleep_summary["TimeInBedHours"] = (
    daily_sleep_summary["TotalTimeInBed"] / 60
)

print("\nDaily Sleep Summary Shape:",
      daily_sleep_summary.shape)

print("\nFirst 10 Records:")
print(daily_sleep_summary.head(10))
```

```python
# Step 7.69 - Heart Rate Analysis

heartrate_seconds["Time"] = pd.to_datetime(
    heartrate_seconds["Time"]
)

# Daily heart-rate summary
daily_heart_rate = (
    heartrate_seconds
    .assign(ActivityDate=heartrate_seconds["Time"].dt.date)
    .groupby(["Id", "ActivityDate"], as_index=False)
    .agg(
        AverageHeartRate=("Value", "mean"),
        MinimumHeartRate=("Value", "min"),
        MaximumHeartRate=("Value", "max"),
        HeartRateReadings=("Value", "count")
    )
)

print("Daily Heart Rate Shape:", daily_heart_rate.shape)
print(daily_heart_rate.head())
```

```python
# Step 7.70 - Weight Analysis

weight_log["Date"] = pd.to_datetime(
    weight_log["Date"]
)

print("Weight Columns:")
print(weight_log.columns.tolist())

print("\nWeight Summary:")
print(weight_log.describe())
```

```python
# Step 7.71 - Main Daily Dataset

daily_activity["ActivityDate"] = pd.to_datetime(
    daily_activity["ActivityDate"]
).dt.date

daily_activity_summary = daily_activity.copy()

print("Daily Activity Shape:",
      daily_activity_summary.shape)

print(daily_activity_summary.head())
```

```python
# Step 7.72 - Merge Daily Analytical Metrics

master_daily = daily_activity_summary.merge(
    daily_calorie_summary,
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDate"],
    how="left"
)

master_daily = master_daily.merge(
    daily_steps_summary,
    on=["Id", "ActivityDate"],
    how="left"
)

master_daily = master_daily.merge(
    daily_intensity,
    on=["Id", "ActivityDate"],
    how="left"
)

print("Master Daily Shape:", master_daily.shape)

print("\nColumns:")
print(master_daily.columns.tolist())

print("\nFirst 5 Rows:")
print(master_daily.head())
```

```python
# Step 7.73 - Master Dataset Quality Check

print("Shape:", master_daily.shape)

print("\nMissing Values:")
print(master_daily.isnull().sum())

print(
    "\nDuplicate Rows:",
    master_daily.duplicated().sum()
)
```

```python
# Step 7.74 - Master Daily Basic Statistics

print("Total Users:",
      master_daily["Id"].nunique())

print(
    "Total Activity Days:",
    master_daily["ActivityDate"].nunique()
)

print(
    "Average Steps:",
    round(
        master_daily["TotalSteps_x"].mean(),
        2
    )
)

print(
    "Average Calories:",
    round(
        master_daily["TotalDailyCalories"].mean(),
        2
    )
)

print(
    "Average Sedentary Minutes:",
    round(
        master_daily["SedentaryMinutes"].mean(),
        2
    )
)

print(
    "Average Distance:",
    round(
        master_daily["TotalDistance"].mean(),
        2
    )
)
```

```python
# Step 7.75 - Activity Level Classification

master_daily["ActivityLevel"] = pd.cut(
    master_daily["TotalSteps_x"],
    bins=[-1, 4999, 9999, float("inf")],
    labels=["Low", "Moderate", "High"]
)

print(master_daily[
    ["Id", "ActivityDate", "TotalSteps_x", "ActivityLevel"]
].head(10))

print("\nActivity Level Distribution:")
print(master_daily["ActivityLevel"].value_counts())
```

```python
# Step 7.76 - Activity Level Analysis

activity_level_summary = (
    master_daily
    .groupby("ActivityLevel", observed=True)
    .agg(
        AverageSteps=("TotalSteps_x", "mean"),
        AverageCalories=("TotalDailyCalories", "mean"),
        AverageDistance=("TotalDistance", "mean"),
        AverageSedentaryMinutes=("SedentaryMinutes", "mean"),
        AverageIntensity=("AverageIntensity", "mean")
    )
    .reset_index()
)

print("Activity Level Summary:")
print(activity_level_summary)
```

```python
# Step 7.77 - Sleep Analysis

print("Average Sleep Hours:",
      round(
          daily_sleep_summary["SleepHours"].mean(),
          2
      ))

print("Average Time in Bed Hours:",
      round(
          daily_sleep_summary["TimeInBedHours"].mean(),
          2
      ))

print("\nSleep Statistics:")
print(
    daily_sleep_summary[
        ["SleepHours", "TimeInBedHours"]
    ].describe()
)
```

```python
# Step 7.78 - Heart Rate Analysis

print("Average Heart Rate:",
      round(
          daily_heart_rate["AverageHeartRate"].mean(),
          2
      ))

print("Minimum Recorded Heart Rate:",
      daily_heart_rate["MinimumHeartRate"].min())

print("Maximum Recorded Heart Rate:",
      daily_heart_rate["MaximumHeartRate"].max())

print("\nHeart Rate Statistics:")
print(
    daily_heart_rate[
        [
            "AverageHeartRate",
            "MinimumHeartRate",
            "MaximumHeartRate"
        ]
    ].describe()
)
```

```python
# Step 7.79 - User-wise Activity Summary

user_activity_summary = (
    master_daily
    .groupby("Id", as_index=False)
    .agg(
        AverageSteps=("TotalSteps_x", "mean"),
        AverageCalories=("TotalDailyCalories", "mean"),
        AverageDistance=("TotalDistance", "mean"),
        AverageSedentaryMinutes=("SedentaryMinutes", "mean"),
        ActivityDays=("ActivityDate", "nunique")
    )
    .sort_values(
        "AverageSteps",
        ascending=False
    )
)

print("User-wise Activity Summary:")
print(user_activity_summary.head(10))
```

```python
# Step 7.80 - Top Active Users

top_active_users = (
    user_activity_summary
    .sort_values(
        "AverageSteps",
        ascending=False
    )
    .head(10)
)

print("Top 10 Active Users:")
print(top_active_users)
```

```python
# Step 7.81 - Correlation Analysis

correlation_columns = [
    "TotalSteps_x",
    "TotalDistance",
    "TotalDailyCalories",
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes",
    "SedentaryMinutes",
    "TotalIntensity",
    "AverageIntensity"
]

correlation_matrix = (
    master_daily[correlation_columns]
    .corr()
)

print("Correlation Matrix:")
print(correlation_matrix)
```

```python
# Step 7.82 - Final Master Dataset Check

print("========== FINAL MASTER DATASET CHECK ==========")

print("\nDataset Shape:")
print(master_daily.shape)

print("\nNumber of Users:")
print(master_daily["Id"].nunique())

print("\nMissing Values:")
print(master_daily.isnull().sum())

print("\nTotal Missing Values:")
print(
    master_daily.isnull().sum().sum()
)

print("\nDuplicate Rows:")
print(
    master_daily.duplicated().sum()
)

print("\nMaster Dataset Columns:")
print(
    master_daily.columns.tolist()
)

print("\n========== ANALYSIS READY ==========")
```
